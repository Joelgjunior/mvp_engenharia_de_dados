```python
import requests

url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.4189/dados"
params = {"formato": "json", "dataInicial": "01/01/2021", "dataFinal": "31/12/2025"}
resp = requests.get(url, params=params)
dados_selic = resp.json()

df_selic = spark.createDataFrame(dados_selic)
df_selic.write.mode("overwrite").saveAsTable("susep_capitalizacao.bronze.selic_mensal")
```
